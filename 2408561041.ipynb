# Latihan 2 : Permen

Kita memiliki lima jenis permen dengan jumlah masing-masing. Pertanyaannya adalah berapa besar peluang mendapatkan Snickers jika satu permen diambil secara acak.

### Chart yang dipilih: Pie Chart

Alasan: Pie chart cocok untuk menunjukkan bagian suatu kategori terhadap keseluruhan. Dengan demikian, bagian Snickers dapat langsung dibandingkan dengan jenis permen lainnya.

```python
import matplotlib.pyplot as plt

candy_names = ['Kit Kat', 'Snickers', 'Milky Way', 'Toblerone', 'Twix']
candy_counts = [52, 39, 78, 13, 78]

plt.figure(figsize=(8, 8))
plt.pie(
    candy_counts,
    labels=candy_names,
    autopct='%1.1f%%',
    startangle=90
)
plt.title('Persentase Jenis Permen')
plt.show()

total_candy = sum(candy_counts)
snickers_probability = candy_counts[1] / total_candy * 100

print(f'Total permen: {total_candy}')
print(f'Peluang memilih Snickers: {snickers_probability:.2f}%')
```

### Jawaban

Jumlah seluruh permen:

52 + 39 + 78 + 13 + 78 = 260

Jumlah Snickers adalah 39.

Peluang memilih Snickers:

39 / 260 × 100% = 15%

Jadi, kemungkinan mendapatkan Snickers adalah 15%.

# Latihan 3 : Makanan

Restoran ingin menghapus tiga makanan penutup yang paling tidak populer. Dataset berisi jumlah penjualan setiap makanan penutup selama satu minggu.

### Chart yang dipilih: Bar Chart

Alasan: Bar chart sangat sesuai untuk membandingkan jumlah penjualan antar makanan penutup. Perbedaan popularitas setiap menu dapat terlihat secara langsung, termasuk tiga menu dengan penjualan paling rendah.

```python
import matplotlib.pyplot as plt

dessert_sales = {
    "Lava Cake": 14,
    "Mousse": 5,
    "Chocolate Cake": 12,
    "Ice Cream": 19,
    "Truffles": 6,
    "Brownie": 8,
    "Chocolate Chip Cookie": 12,
    "Chocolate Pudding": 9,
    "Souffle": 10,
    "Chocolate Cheesecake": 17,
    "Chocolate Chips": 2,
    "Fudge": 9,
    "Mochi": 13,
}

# Urutkan dari penjualan paling sedikit ke paling banyak
sorted_sales = sorted(dessert_sales.items(), key=lambda item: item[1])
desserts = [item[0] for item in sorted_sales]
sales = [item[1] for item in sorted_sales]

plt.figure(figsize=(12, 7))
plt.barh(desserts, sales)
plt.title('Popularitas Makanan Penutup Berdasarkan Penjualan')
plt.xlabel('Jumlah Terjual dalam Seminggu')
plt.ylabel('Makanan Penutup')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()

least_popular = sorted_sales[:3]
print('Tiga makanan penutup dengan penjualan paling rendah:')
for dessert, count in least_popular:
    print(f'- {dessert}: {count} terjual')
```

### Jawaban

Tiga makanan penutup dengan jumlah penjualan paling rendah adalah:

1. Chocolate Chips : 2 terjual
2. Mousse : 5 terjual
3. Truffles : 6 terjual

Jadi, berdasarkan data penjualan, ketiga makanan tersebut merupakan tiga menu dengan popularitas paling rendah.

# Latihan 4 : Penggunaan CPU

Data menunjukkan penggunaan CPU rata-rata setiap jam selama tujuh hari, mulai dari Senin sampai Minggu.

Pertanyaan yang harus dijawab:

- Jam berapa pekerja biasanya makan siang?
- Apakah pekerja bekerja pada akhir pekan?
- Pada hari apa pekerja mulai bekerja pada komputer mereka pada malam hari?

### Chart yang dipilih: Heatmap

Alasan: Heatmap sangat sesuai karena data memiliki dua dimensi, yaitu hari dan jam. Intensitas warna dapat menunjukkan tinggi-rendahnya penggunaan CPU sehingga pola aktivitas selama satu minggu dapat terlihat dengan mudah.

```python
import matplotlib.pyplot as plt
import numpy as np

cpu_usage = [
    [2, 2, 4, 2, 4, 1, 1, 4, 4, 12, 22, 23, 45, 9, 33, 56, 23, 40, 21, 6, 6, 2, 2, 3],
    [1, 2, 3, 2, 3, 2, 3, 2, 7, 22, 45, 44, 33, 9, 23, 19, 33, 56, 12, 2, 3, 1, 2, 2],
    [2, 3, 1, 2, 4, 4, 2, 2, 1, 2, 5, 31, 54, 7, 6, 34, 68, 34, 49, 6, 6, 2, 2, 3],
    [1, 2, 3, 2, 4, 1, 2, 4, 1, 17, 24, 18, 41, 3, 44, 42, 12, 36, 41, 2, 2, 4, 2, 4],
    [4, 1, 2, 2, 3, 2, 5, 1, 2, 12, 33, 27, 43, 8, 38, 53, 29, 45, 39, 3, 1, 1, 3, 4],
    [2, 3, 1, 2, 2, 5, 2, 8, 4, 2, 3, 1, 5, 1, 2, 3, 2, 6, 1, 2, 2, 1, 4, 3],
    [1, 2, 3, 1, 1, 3, 4, 2, 3, 1, 2, 2, 5, 3, 2, 1, 4, 2, 45, 26, 33, 2, 2, 1],
]

days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
hours = list(range(24))

plt.figure(figsize=(14, 6))
plt.imshow(cpu_usage, aspect='auto')
plt.colorbar(label='CPU Usage (%)')
plt.xticks(hours)
plt.yticks(range(7), days)
plt.xlabel('Jam')
plt.ylabel('Hari')
plt.title('Heatmap Penggunaan CPU Selama Seminggu')
plt.tight_layout()
plt.show()
```

```python
# Analisis sederhana berdasarkan dataset

cpu = np.array(cpu_usage)

# Rata-rata penggunaan CPU pada setiap jam untuk hari kerja
weekday_avg = cpu[:5].mean(axis=0)

# Cari jam dengan penggunaan CPU paling rendah di sekitar periode kerja
print('Rata-rata penggunaan CPU per jam pada Senin-Jumat:')
for hour, value in enumerate(weekday_avg):
    print(f'Jam {hour:02d}:00 = {value:.2f}%')

# Penggunaan CPU akhir pekan
print('\nPenggunaan CPU maksimum pada Sabtu:', cpu[5].max(), '%')
print('Penggunaan CPU maksimum pada Minggu:', cpu[6].max(), '%')

# Aktivitas malam (18:00-23:00)
print('\nPenggunaan CPU malam hari (18:00-23:00):')
for i, day in enumerate(days):
    evening_max = cpu[i, 18:24].max()
    print(f'{day}: maksimum {evening_max}%')
```

### Jawaban

1. Jam berapa pekerja biasanya makan siang?

Dari heatmap, penggunaan CPU pada sekitar jam 13:00 cenderung turun pada hari kerja. Hal ini menunjukkan bahwa pekerja kemungkinan sedang makan siang pada sekitar pukul 13.00.

2. Apakah pekerja bekerja pada akhir pekan?

Pada Sabtu, penggunaan CPU relatif rendah sepanjang hari sehingga tidak terlihat aktivitas kerja yang berarti.

Pada Minggu, terdapat peningkatan penggunaan CPU yang cukup tinggi sekitar pukul 18.00–20.00, terutama pada pukul 18.00 dengan nilai 45%.

Jadi, berdasarkan dataset, terdapat aktivitas komputer pada akhir pekan, terutama pada Minggu malam.

3. Pada hari apa pekerja mulai bekerja pada komputer mereka pada malam hari?

Aktivitas malam yang paling menonjol terdapat pada hari Minggu, ketika penggunaan CPU meningkat menjadi 45% pada pukul 18.00 dan tetap cukup tinggi setelahnya.
